```python
# ==========================================
# CREATE SERVICE DATES FROM GTFS CALENDAR
# ==========================================

import pandas as pd

# Convert GTFS dates to real dates
calendar["start_date"] = pd.to_datetime(
    calendar["start_date"].astype(str),
    format="%Y%m%d"
)

calendar["end_date"] = pd.to_datetime(
    calendar["end_date"].astype(str),
    format="%Y%m%d"
)

calendar_dates["date"] = pd.to_datetime(
    calendar_dates["date"].astype(str),
    format="%Y%m%d"
)

# Create all dates between start_date and end_date
service_dates = []

for _, row in calendar.iterrows():

    dates = pd.date_range(
        start=row["start_date"],
        end=row["end_date"],
        freq="D"
    )

    for date in dates:

        weekday = date.day_name().lower()

        if row[weekday] == 1:
            service_dates.append({
                "service_id": row["service_id"],
                "service_date": date
            })

# Convert to DataFrame
service_dates = pd.DataFrame(service_dates)

print("Service dates created successfully!")
print("Rows:", len(service_dates))

display(service_dates.head(10))
```

```python
# ==========================================
# APPLY GTFS SERVICE EXCEPTIONS
# ==========================================

# Separate added and removed services
added_services = calendar_dates[
    calendar_dates["exception_type"] == 1
][["service_id", "date"]].rename(
    columns={"date": "service_date"}
)

removed_services = calendar_dates[
    calendar_dates["exception_type"] == 2
][["service_id", "date"]].rename(
    columns={"date": "service_date"}
)

# Remove services that are explicitly cancelled
service_dates = service_dates.merge(
    removed_services.assign(remove_flag=1),
    on=["service_id", "service_date"],
    how="left"
)

service_dates = service_dates[
    service_dates["remove_flag"].isna()
].drop(columns=["remove_flag"])

# Add services that were explicitly added
service_dates = pd.concat(
    [service_dates, added_services],
    ignore_index=True
)

# Remove any accidental duplicates
service_dates = service_dates.drop_duplicates(
    subset=["service_id", "service_date"]
).sort_values(
    ["service_id", "service_date"]
).reset_index(drop=True)

print("===== FINAL SERVICE DATES =====")
print("Rows:", len(service_dates))
print(
    "Date range:",
    service_dates["service_date"].min(),
    "to",
    service_dates["service_date"].max()
)

display(service_dates.head(10))
```

```python
# ==========================================
# CREATE TRIP-DATE SCHEDULE
# ==========================================

scheduled_trips = trips.merge(
    service_dates,
    on="service_id",
    how="inner"
)

print("===== SCHEDULED TRIPS =====")
print("Rows:", len(scheduled_trips))
print("Columns:", scheduled_trips.columns.tolist())

display(scheduled_trips.head(10))
```

```python
# ==========================================
# VALIDATE TRIP-DATE DATASET
# ==========================================

print("Unique trips:", scheduled_trips["trip_id"].nunique())
print("Unique service dates:", scheduled_trips["service_date"].nunique())

print(
    "Date range:",
    scheduled_trips["service_date"].min(),
    "to",
    scheduled_trips["service_date"].max()
)

print(
    "Missing service dates:",
    scheduled_trips["service_date"].isna().sum()
)

print(
    "Missing trip IDs:",
    scheduled_trips["trip_id"].isna().sum()
)
```

```python
# ==========================================
# ADD ROUTE INFORMATION
# ==========================================

scheduled_trips = scheduled_trips.merge(
    routes[
        [
            "route_id",
            "route_short_name",
            "route_long_name",
            "route_type"
        ]
    ],
    on="route_id",
    how="left"
)

print("Final scheduled trip columns:")
print(scheduled_trips.columns.tolist())

display(
    scheduled_trips[
        [
            "service_date",
            "route_id",
            "route_short_name",
            "route_long_name",
            "trip_id",
            "direction"
        ]
    ].head(20)
)
```

```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
GTFS_DIR = PROJECT_ROOT / "data" / "raw" / "GTFS"

routes = pd.read_csv(GTFS_DIR / "routes.txt")
trips = pd.read_csv(GTFS_DIR / "trips.txt")
calendar = pd.read_csv(GTFS_DIR / "calendar.txt")
calendar_dates = pd.read_csv(GTFS_DIR / "calendar_dates.txt")

print("Routes:", len(routes))
print("Trips:", len(trips))
print("Calendar rows:", len(calendar))
print("Calendar exceptions:", len(calendar_dates))
```

```python

# Convert GTFS dates from YYYYMMDD to datetime
calendar["start_date"] = pd.to_datetime(
    calendar["start_date"].astype(str), format="%Y%m%d"
)
calendar["end_date"] = pd.to_datetime(
    calendar["end_date"].astype(str), format="%Y%m%d"
)
calendar_dates["date"] = pd.to_datetime(
    calendar_dates["date"].astype(str), format="%Y%m%d"
)

# Generate dates on which each regular service operates
service_dates_list = []

weekday_columns = [
    "monday", "tuesday", "wednesday", "thursday",
    "friday", "saturday", "sunday"
]

for _, row in calendar.iterrows():
    dates = pd.date_range(
        start=row["start_date"],
        end=row["end_date"],
        freq="D"
    )

    for date in dates:
        weekday = date.day_name().lower()

        if row[weekday] == 1:
            service_dates_list.append({
                "service_id": row["service_id"],
                "service_date": date
            })

service_dates = pd.DataFrame(service_dates_list)

# Apply exceptions: 1 = service added, 2 = service removed
added = calendar_dates.loc[
    calendar_dates["exception_type"] == 1,
    ["service_id", "date"]
].rename(columns={"date": "service_date"})

removed = calendar_dates.loc[
    calendar_dates["exception_type"] == 2,
    ["service_id", "date"]
].rename(columns={"date": "service_date"})

service_dates = service_dates.merge(
    removed.assign(remove_flag=True),
    on=["service_id", "service_date"],
    how="left"
)

service_dates = service_dates[
    service_dates["remove_flag"].isna()
].drop(columns="remove_flag")

service_dates = pd.concat(
    [service_dates, added],
    ignore_index=True
)

service_dates = (
    service_dates
    .drop_duplicates(["service_id", "service_date"])
    .sort_values(["service_id", "service_date"])
    .reset_index(drop=True)
)

print("Service-date rows:", len(service_dates))
print("First service date:", service_dates["service_date"].min())
print("Last service date:", service_dates["service_date"].max())
print("Unique service IDs:", service_dates["service_id"].nunique())
```

```python

# Attach operating dates to trips
scheduled_trips = trips.merge(
    service_dates,
    on="service_id",
    how="inner"
)

# Attach route details
route_columns = [
    "route_id",
    "route_short_name",
    "route_long_name",
    "route_type"
]

scheduled_trips = scheduled_trips.merge(
    routes[route_columns],
    on="route_id",
    how="left",
    validate="many_to_one"
)

# Validate and summarize
print("Scheduled trip-date rows:", len(scheduled_trips))
print("Unique trips:", scheduled_trips["trip_id"].nunique())
print("Unique service dates:", scheduled_trips["service_date"].nunique())
print("Missing route names:", scheduled_trips["route_long_name"].isna().sum())
print(
    "Duplicate trip-date pairs:",
    scheduled_trips.duplicated(["trip_id", "service_date"]).sum()
)

display(scheduled_trips.head())
```

```python

# Create a route-by-date summary
route_daily = (
    scheduled_trips
    .groupby(
        [
            "service_date",
            "route_id",
            "route_short_name",
            "route_long_name",
            "route_type"
        ],
        dropna=False,
        as_index=False
    )
    .agg(scheduled_trips=("trip_id", "nunique"))
)

route_daily["service_date"] = pd.to_datetime(
    route_daily["service_date"]
)

route_daily["day_of_week"] = (
    route_daily["service_date"].dt.day_name()
)

route_daily["route_name"] = (
    route_daily["route_short_name"]
    .fillna(route_daily["route_long_name"])
    .astype(str)
)

print("Route-date summary rows:", len(route_daily))
print("Unique routes:", route_daily["route_id"].nunique())
print("Unique dates:", route_daily["service_date"].nunique())

display(
    route_daily.sort_values(
        ["service_date", "scheduled_trips"],
        ascending=[True, False]
    ).head(10)
)
```

```python
OUTPUT_DIR = PROJECT_ROOT / "data" / "processed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

output_file = OUTPUT_DIR / "route_daily_scheduled_trips.csv"
route_daily.to_csv(output_file, index=False)

print("Saved:", output_file)
print("File size (MB):", round(output_file.stat().st_size / (1024**2), 2))
```

```python

# Rank routes by average scheduled trips per service date
route_ranking = (
    route_daily
    .groupby(
        ["route_id", "route_name", "route_long_name", "route_type"],
        dropna=False,
        as_index=False
    )
    .agg(
        total_scheduled_trips=("scheduled_trips", "sum"),
        active_service_days=("service_date", "nunique"),
        average_trips_per_service_day=("scheduled_trips", "mean")
    )
)

route_ranking["average_trips_per_service_day"] = (
    route_ranking["average_trips_per_service_day"].round(2)
)

route_ranking = route_ranking.sort_values(
    "average_trips_per_service_day",
    ascending=False
)

display(route_ranking.head(15))
```

```python

rolling_calendar_query = """
WITH date_spine AS (
    SELECT CAST(d AS DATE) AS service_date
    FROM generate_series(
        (SELECT MIN(service_date) FROM route_daily),
        (SELECT MAX(service_date) FROM route_daily),
        INTERVAL 1 DAY
    ) AS dates(d)
),
route_list AS (
    SELECT DISTINCT
        route_id,
        route_name,
        route_long_name
    FROM route_daily
),
daily_filled AS (
    SELECT
        r.route_id,
        r.route_name,
        r.route_long_name,
        d.service_date,
        COALESCE(rd.scheduled_trips, 0) AS scheduled_trips
    FROM route_list r
    CROSS JOIN date_spine d
    LEFT JOIN route_daily rd
        ON rd.route_id = r.route_id
        AND rd.service_date = d.service_date
),
rolling AS (
    SELECT
        *,
        AVG(scheduled_trips) OVER (
            PARTITION BY route_id
            ORDER BY service_date
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        ) AS rolling_7_calendar_day_average
    FROM daily_filled
)
SELECT
    route_id,
    route_name,
    service_date,
    scheduled_trips,
    ROUND(rolling_7_calendar_day_average, 2)
        AS rolling_7_calendar_day_average
FROM rolling
ORDER BY route_id, service_date
"""

rolling_calendar_results = con.execute(
    rolling_calendar_query
).df()

display(rolling_calendar_results.head(15))
```

```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

# Normalize the keys so both datasets use the same types
scheduled_trips["service_date"] = pd.to_datetime(
    scheduled_trips["service_date"]
).dt.normalize()

scheduled_trips["service_id"] = (
    scheduled_trips["service_id"].astype(str).str.strip()
)

service_sets = {
    date: {str(service_id).strip() for service_id in ids}
    for date, ids in service_sets.items()
}

dates_to_check = pd.to_datetime([
    "2026-08-31",
    "2026-09-07",
    "2026-09-14"
])

daily_service_counts = (
    scheduled_trips[
        scheduled_trips["service_date"].isin(dates_to_check)
    ]
    .groupby(["service_date", "service_id"], as_index=False)
    .agg(scheduled_trips=("trip_id", "nunique"))
)

all_rows = []

for date in dates_to_check:
    active_ids = service_sets[date]

    date_counts = (
        daily_service_counts[
            daily_service_counts["service_date"] == date
        ]
        .set_index("service_id")["scheduled_trips"]
        .to_dict()
    )

    for service_id in sorted(active_ids):
        all_rows.append({
            "service_date": date,
            "service_id": service_id,
            "scheduled_trips": date_counts.get(service_id, 0)
        })

service_date_counts = pd.DataFrame(all_rows)

summary = (
    service_date_counts
    .groupby("service_date", as_index=False)
    .agg(
        active_service_ids=("service_id", "nunique"),
        scheduled_trips=("scheduled_trips", "sum"),
        service_ids_with_trips=(
            "scheduled_trips",
            lambda values: (values > 0).sum()
        )
    )
)

print(summary.to_string(index=False))

print("\nValidation against the original network totals:")
print(
    scheduled_trips[
        scheduled_trips["service_date"].isin(dates_to_check)
    ]
    .groupby("service_date")["trip_id"]
    .nunique()
    .to_string()
)

service_date_counts.to_csv(
    PROCESSED_DIR / "scheduled_trips_by_service_id_and_date.csv",
    index=False
)

print("\nSaved corrected service-level analysis.")
```

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

route_daily = pd.read_csv(
    PROCESSED_DIR / "route_daily_scheduled_trips.csv",
    parse_dates=["service_date"]
)

# Aggregate route-level records into network-level daily metrics
network_daily = (
    route_daily.groupby("service_date", as_index=False)
    .agg(
        network_scheduled_trips=("scheduled_trips", "sum"),
        routes_with_service=("scheduled_trips", lambda s: (s > 0).sum()),
        routes_without_service=("scheduled_trips", lambda s: (s == 0).sum())
    )
)

network_daily["day_of_week"] = (
    network_daily["service_date"].dt.day_name()
)

network_daily["is_weekend"] = (
    network_daily["service_date"].dt.dayofweek >= 5
)

# Add a simple seven-calendar-day rolling average
network_daily = network_daily.sort_values("service_date")

network_daily["rolling_7day_average"] = (
    network_daily["network_scheduled_trips"]
    .rolling(window=7, min_periods=1)
    .mean()
    .round(2)
)

output = PROCESSED_DIR / "network_daily_dashboard.csv"
network_daily.to_csv(output, index=False)

print("Saved:", output)
print("Rows:", len(network_daily))
print("\nSample:")
print(network_daily.head(10).to_string(index=False))

print("\nLowest-service dates:")
print(
    network_daily.nsmallest(10, "network_scheduled_trips")
    .to_string(index=False)
)
```
